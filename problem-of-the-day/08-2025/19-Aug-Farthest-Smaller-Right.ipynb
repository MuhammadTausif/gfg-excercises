## Farthest Smaller Right

You are given an array arr[]. For each element at index i (0-based indexing), find the farthest index j to the right (i.e., j > i) such that arr[j] < arr[i]. If no such index exists for a given position, return -1 for that index. Return the resulting array of answers.

### Examples:

---
- Input: arr[] = [2, 5, 1, 3, 2]
- Output: [2, 4, -1, 4, -1]
- Explanation: arr[0] = 2: Farthest smaller element to the right is arr[2] = 1.
arr[1] = 5: Farthest smaller element to the right is arr[4] = 2.
arr[2] = 1: No smaller element to the right → -1.
arr[3] = 3: Farthest smaller element to the right is arr[4] = 2.
arr[4] = 2: No elements to the right → -1.

---
- Input: arr[] = [2, 3, 5, 4, 1] 
- Output: [4, 4, 4, 4, -1]
- Explanation: arr[4] is the farthest smallest element to the right for arr[0], arr[1], arr[2] and arr[3].

```python
class Solution:
    def farMin(self, arr):
        # Code Here
        # credit to @Mateusz Dereniowski
        # sort the index by the arr[i] value, then at any "sorted" index (by value),
        # we have seen all the indexes whose value <= value of current index (because of the equal
        # value, you have to make sure the sort is stable sort. for same value, smaller index is in 
        # front, otherwise, it will break 
        # we keep the largest "seen" index and if it is bigger than the current index,
        # which means it is on the right side of current index, then it is the answer
        # for the index
        n = len(arr)
        indices = sorted(list(range(n)), key=lambda i: arr[i])
        #print(f"{indices = }")
        largest_index = indices[0] # the largest index that we have seen (have seen means all indexes whoose value < value of current index)
        ans = [0]*n
        
        for i in indices:
            if largest_index > i: #means largest_index is on the right side of i
                ans[i] = largest_index
            else: # there is no smaller value that is on the right of this index.
                ans[i] = -1
            largest_index = max(largest_index, i)
        
        return ans
```

```python
arr = [[2, 5, 1, 3, 2], [2, 3, 5, 4, 1] ]
sol = Solution()
for a in arr:
    print(f"Input: arr[] = {a}")
    print(f"Output: {sol.farMin(a)}")
    print("---"*10)
```
